````python
from torchvision.transforms.functional import crop
from tqdm import tqdm
import supervision as sv

STRIDE = 30
PLAYER_ID = 2

def extract_crops(video_path):
    frame_generator = sv.get_video_frames_generator(INPUT_VIDEO, stride=STRIDE)

    crops = []
    for frame in tqdm(frame_generator, desc='Collecting crops'):
      result = model.predict(frame, conf=0.3)[0]
      detections = sv.Detections.from_ultralytics(result)
      detections = detections.with_nms(threshold=0.5, class_agnostic=True)
      detections = detections[detections.class_id == PLAYER_ID]
      crops += [
          sv.crop_image(frame, xyxy)
          for xyxy in detections.xyxy
      ]
    return crops
````

````python
import torch
from transformers import AutoProcessor, SiglipVisionModel

SIGLIP_MODEL_PATH = 'google/siglip-base-patch16-224'
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
EMBEDDINGS_MODEL = SiglipVisionModel.from_pretrained(SIGLIP_MODEL_PATH).to(DEVICE)
EMBEDDINGS_PROCESSOR = AutoProcessor.from_pretrained(SIGLIP_MODEL_PATH)
````

````python
from more_itertools import chunked
import numpy as np

BATCH_SIZE = 32

crops = [sv.cv2_to_pillow(crop) for crop in crops]
batches = chunked(crops, BATCH_SIZE)
data = []

with torch.no_grad():
    for batch in tqdm(batches, desc='Embeddings Extractions'):
        inputs = EMBEDDINGS_PROCESSOR(images=batch, return_tensors='pt').to(DEVICE)
        outputs = EMBEDDINGS_MODEL(**inputs)
        embeddings = torch.mean(outputs.last_hidden_state, dim=1).cpu().detach().numpy()
        data.append(embeddings)

data = np.concatenate(data)
````

````python
import os
import cv2
import supervision as sv
from tqdm import tqdm

STRIDE = 30
PLAYER_ID = 2

VALIDATION_CROP_DIR = "/content/team_validation_crops"
os.makedirs(VALIDATION_CROP_DIR, exist_ok=True)


def extract_player_crops_with_metadata(video_path):

    frame_generator = sv.get_video_frames_generator(
        video_path,
        stride=STRIDE
    )

    crop_data = []

    for frame_idx, frame in enumerate(
        tqdm(frame_generator, desc="Extracting player crops")
    ):

        result = model.predict(
            frame,
            conf=0.3,
            verbose=False
        )[0]

        detections = sv.Detections.from_ultralytics(result)

        detections = detections.with_nms(
            threshold=0.5,
            class_agnostic=True
        )

        # Keep only players
        detections = detections[
            detections.class_id == PLAYER_ID
        ]

        for detection_idx, xyxy in enumerate(detections.xyxy):

            crop = sv.crop_image(frame, xyxy)

            crop_path = os.path.join(
                VALIDATION_CROP_DIR,
                f"frame_{frame_idx:05d}_player_{detection_idx:02d}.jpg"
            )

            cv2.imwrite(
                crop_path,
                cv2.cvtColor(crop, cv2.COLOR_RGB2BGR)
            )

            crop_data.append({
                "frame_idx": frame_idx,
                "detection_idx": detection_idx,
                "path": crop_path
            })

    return crop_data
````

````python
import torch
import numpy as np
import pandas as pd
from PIL import Image
from tqdm import tqdm
from more_itertools import chunked

# ============================================================
# SIGLIP
# ============================================================

SIGLIP_MODEL_PATH = "google/siglip-base-patch16-224"

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

EMBEDDINGS_MODEL = SiglipVisionModel.from_pretrained(
    SIGLIP_MODEL_PATH
).to(DEVICE)

EMBEDDINGS_MODEL.eval()

EMBEDDINGS_PROCESSOR = AutoProcessor.from_pretrained(
    SIGLIP_MODEL_PATH
)

BATCH_SIZE = 32


# ============================================================
# EMBEDDING FUNCTION
# ============================================================

def extract_embeddings(paths, batch_size=32):

    embeddings_list = []

    for batch_paths in tqdm(
        list(chunked(paths, batch_size)),
        desc="Extracting SigLIP embeddings"
    ):

        images = []

        for path in batch_paths:
            image = Image.open(path).convert("RGB")
            images.append(image)

        inputs = EMBEDDINGS_PROCESSOR(
            images=images,
            return_tensors="pt"
        )

        inputs = {
            key: value.to(DEVICE)
            for key, value in inputs.items()
        }

        with torch.no_grad():

            outputs = EMBEDDINGS_MODEL(**inputs)

            embeddings = torch.mean(
                outputs.last_hidden_state,
                dim=1
            )

        embeddings_list.append(
            embeddings.cpu().numpy()
        )

    return np.concatenate(embeddings_list, axis=0)
````

````python
development_paths = development_df["path"].tolist()

development_embeddings = extract_embeddings(
    development_paths
)

print(
    "Development embeddings shape:",
    development_embeddings.shape
)
````

# 📝 Team Classification Validation

## Objective

The objective of this validation is to quantitatively evaluate the **team classification component** of the football analytics system.

The implemented approach uses:

**SigLIP → UMAP → KMeans**

to group detected football players into two teams based on their visual appearance.

The validation is performed separately from the main football-analysis pipeline using manually labelled player crops as ground truth.

---

## 1. Player Crop Extraction

Player crops were extracted from the football match video using the trained YOLO object-detection model.

Only detections belonging to the **player class (`PLAYER_ID = 2`)** were retained.

A frame stride of **30** was used to sample the video.

Each detected player crop was saved along with its metadata:

* Frame index
* Detection index
* Crop image path

The extraction process produced approximately **500 player crops** from the video.

For the validation experiment, **200 player crops** were selected for manual annotation.

---

## 2. Ground-Truth Labelling

The selected player crops were manually labelled according to their actual team based on the player's jersey colour.

The team labels were defined as:

* **Team A → Yellow**
* **Team B → White**

A total of **200 player crops** were manually labelled.

| Team            | Number of Samples |
| --------------- | ----------------: |
| Team A — Yellow |               101 |
| Team B — White  |                99 |
| **Total**       |           **200** |

No samples were skipped during the labelling process.

The resulting ground-truth labels were stored in:

```text
/content/team_ground_truth.json
```

---

## 3. Development and Test Split

Since the player crops were extracted from a video, a random split was avoided.

A **temporal split** was used to reduce the possibility of data leakage caused by visually similar player crops from nearby video frames appearing in both development and test sets.

The data was divided approximately as:

* **70% → Development set**
* **30% → Test set**

### Dataset Distribution

| Dataset     |   Total |  Yellow |  White |
| ----------- | ------: | ------: | -----: |
| Development |     141 |      76 |     65 |
| Test        |      59 |      25 |     34 |
| **Total**   | **200** | **101** | **99** |

The development set was used to fit the UMAP dimensionality-reduction model and KMeans clustering model.

The test set was kept unseen during fitting and was used only for final evaluation.

---

# 4. SigLIP Feature Extraction

The **SigLIP vision model** was used as a pretrained visual feature extractor.

Model:

```text
google/siglip-base-patch16-224
```

Each player crop was converted into a visual embedding using SigLIP.

Each crop produced a **768-dimensional feature vector**.

Therefore:

### Development set

```text
141 player crops × 768 features
```

Result:

```text
(141, 768)
```

### Test set

```text
59 player crops × 768 features
```

Result:

```text
(59, 768)
```

The SigLIP embeddings capture visual information from the player crops, which can then be used to distinguish players belonging to different teams.

---

# 5. UMAP Dimensionality Reduction

The 768-dimensional SigLIP embeddings were reduced using **UMAP (Uniform Manifold Approximation and Projection)**.

Configuration:

```python
UMAP(
    n_components=3,
    random_state=42
)
```

The dimensionality was reduced from:

```text
768 dimensions
        ↓
3 dimensions
```

For the development set:

```text
(141, 768) → (141, 3)
```

UMAP was **fitted only on the development data**.

The test data was later transformed using the already-fitted UMAP model.

---

# 6. KMeans Clustering

After UMAP dimensionality reduction, **KMeans clustering** was used to divide the player embeddings into two clusters.

Configuration:

```python
KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)
```

The two clusters represent the two opposing teams.

The KMeans model was fitted **only on the development set**.

The resulting development-set cluster distribution was:

| Cluster   | Samples |
| --------- | ------: |
| Cluster 0 |      89 |
| Cluster 1 |      52 |
| **Total** | **141** |

---

# 7. Cluster-to-Team Mapping

Because KMeans is an unsupervised clustering algorithm, its cluster labels (`0` and `1`) do not inherently correspond to Team A or Team B.

The development-set ground-truth labels were therefore used to determine the mapping between clusters and actual teams.

The resulting mapping was:

```text
Cluster 0 → Team A → Yellow
Cluster 1 → Team B → White
```

```text
128 / 141 = 90.78%
```

This value is considered a **development diagnostic**, not the final model performance.
